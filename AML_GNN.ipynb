```python
import torch
import matplotlib.pyplot as plt
import os

X = torch.tensor([
    [0.9],
    [0.10],
    [0.05]
])

# Create a label - "Bad guy - LimFW" 1, Good guy = "Owen 0"
y = torch.tensor([
    1,
    1,
    0
])

# Create the transaction network
A = torch.tensor([
    [0,1,0],
    [0,0,1],
    [0,0,0]
], dtype = torch.float)

I = torch.eye(3) # Create Identity Matrix
A_hat = A + I

print(A_hat)

# Normalise it
degree = A_hat.sum(dim = 1)
D = torch.diag(degree ** - 0.5)
A_norm = D @ A_hat @ D

# Manually create the weights - actual (random)
w = torch.tensor([
    [0.5, 0.6]
])
H = A_norm @ X

# Prediction
Output = H @ w

# Go Github website - Create a repository - AML_GNN
print(os.getcwd())
```
